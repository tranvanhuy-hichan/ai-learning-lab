# Model study on Kaggle (2 x T4)
Runs every heavy training job of `projects/model-study` and zips the results.

**Before you run it**, in the right-hand *Session options*: set **Accelerator = GPU T4 x2** and **Internet = On** (datasets are downloaded).

What it does (roughly 1.5 hours on 2 x T4; each stage prints its progress):
1. downloads the waste and Oxford-IIIT Pet datasets,
2. ablation: 5 training recipes x 3 seeds, split across the 2 GPUs,
3. distillation students (small CNN and MobileNetV3-Small, with and without a teacher),
4. post-training compression: ONNX, int8 quantisation, pruning,
5. segmentation ablation (pretrained encoder, recipe, border weight, resolution),
6. zips `results/` into `/kaggle/working/model-study-results.zip`.

Use *Save Version > Save & Run All*, then download the zip from the **Output** tab. To skip the segmentation stage add `--skip-seg` to the command in the last cell.
